```python
import sympy as sp

x = sp.Symbol('x')

f = x**3 - 6 * x**2 + 9 * x + 1

print("Function: ")
print(f)
f_prime = sp.diff(f, x)

print("\nFirst Derivative: ")
print(f_prime)

critical_point = sp.solve(f_prime, x)

print("\nSecond Derivtive")
f_double = sp.diff(f_prime, x)

print("\nCritical point analysis: ")

for point in critical_point:
    second_value = f_double.subs(x, point)
    if second_value > 0:
        print(f"x = {point} : Local Minimun")
    elif second_value < 0 :
        print(f"x = {point} : Local Maximum")
    else :
        print(f"x = {point} : Test inconclusive")
```
